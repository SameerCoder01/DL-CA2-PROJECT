```
from pathlib import Path
import subprocess, sys
run = Path("/content/drive/MyDrive/DL-CA2-PROJECT/pix2pix")
command = [sys.executable, "model.py", "--epochs", "3", "--samples", "800", "--batch-size", "4", "--output", str(run)]
if (run / "last.pt").exists():
    command += ["--resume", str(run / "last.pt")]
subprocess.run(command, check=True)
```

```
subprocess.run([sys.executable, "evaluate.py", "--checkpoint", str(run / "best.pt"), "--output", str(run / "evaluation")], check=True)
import pandas as pd
history = pd.read_csv(run / "history.csv")
history.plot(x="epoch", y=["psnr", "input_psnr"], title="Held-out synthetic validation PSNR")
```
